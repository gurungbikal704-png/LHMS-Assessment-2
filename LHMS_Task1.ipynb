```python
# LANGHAM HOTEL MANAGEMENT SYSTEM
# Task 1 - Part A and Part B

import os
from datetime import datetime

# Initialising dictionaries
rooms = {}
allocations = {}

# File name
filename = "LHMS_850011021.txt"


# MENU OPTION 1 - ADD ROOM

def add_room():
    print("\n--- ADD ROOM ---")

    room_number = input("Enter Room Number: ")

    if room_number in rooms:
        print("Room already exists!")
        return

    room_type = input("Enter Room Type: ")
    price = float(input("Enter Price Per Night: "))
    features = input("Enter Room Features: ")

    rooms[room_number] = {
        "type": room_type,
        "price": price,
        "features": features,
        "status": "Available"
    }

    print("Room added successfully!")


# MENU OPTION 2 - DELETE ROOM

def delete_room():
    print("\n--- DELETE ROOM ---")

    room_number = input("Enter Room Number to Delete: ")

    if room_number in rooms:

        if rooms[room_number]["status"] == "Available":
            del rooms[room_number]
            print("Room deleted successfully!")

        else:
            print("Cannot delete room. Room is currently allocated.")

    else:
        print("Room not found!")


# MENU OPTION 3 - DISPLAY ROOM DETAILS

def display_rooms():
    print("\n--- ROOM DETAILS ---")

    if not rooms:
        print("No rooms available in the system.")
        return

    for room_number, details in rooms.items():

        print("\nRoom Number:", room_number)
        print("Room Type:", details["type"])
        print("Price Per Night: $", details["price"])
        print("Features:", details["features"])
        print("Status:", details["status"])


# MENU OPTION 4 - ALLOCATE ROOMS

def allocate_room():
    print("\n--- ALLOCATE ROOM ---")

    room_number = input("Enter Room Number: ")

    if room_number not in rooms:
        print("Room does not exist!")
        return

    if rooms[room_number]["status"] == "Allocated":
        print("Room is already allocated!")
        return

    customer_name = input("Enter Customer Name: ")
    nights = int(input("Enter Number of Nights: "))

    if nights <= 0:
        print("Number of nights must be greater than zero.")
        return

    allocations[room_number] = {
        "customer": customer_name,
        "nights": nights,
        "date": datetime.now().strftime("%Y-%m-%d")
    }

    rooms[room_number]["status"] = "Allocated"

    print("Room allocated successfully!")


# MENU OPTION 5 - DISPLAY ROOM ALLOCATION DETAILS

def display_allocations():
    print("\n--- ROOM ALLOCATION DETAILS ---")

    if not allocations:
        print("No rooms are currently allocated.")
        return

    for room_number, details in allocations.items():

        print("\nRoom Number:", room_number)
        print("Customer Name:", details["customer"])
        print("Number of Nights:", details["nights"])
        print("Allocation Date:", details["date"])
        print("Room Status:", rooms[room_number]["status"])


# MENU OPTION 6 - BILLING AND DE-ALLOCATION

def billing():
    print("\n--- BILLING AND DE-ALLOCATION ---")

    room_number = input("Enter Room Number: ")

    if room_number not in allocations:
        print("Room is not currently allocated!")
        return

    customer = allocations[room_number]["customer"]
    nights = allocations[room_number]["nights"]
    price = rooms[room_number]["price"]

    total_bill = nights * price

    print("\n--- CUSTOMER BILL ---")
    print("Customer Name:", customer)
    print("Room Number:", room_number)
    print("Number of Nights:", nights)
    print("Price Per Night: $", price)
    print("Total Bill: $", total_bill)

    print("\nPayment completed.")

    del allocations[room_number]

    rooms[room_number]["status"] = "Available"

    print("Room de-allocated successfully!")
    print("Room is now available for the next booking.")


# MENU OPTION 7 - SAVE ROOM ALLOCATION TO FILE

def save_to_file():
    print("\n--- SAVE ROOM ALLOCATION ---")

    try:
        with open(filename, "w") as file:

            file.write("LANGHAM HOTEL MANAGEMENT SYSTEM\n")
            file.write("Room Details and Allocation Records\n")
            file.write("=" * 40 + "\n")

            file.write("\nROOM DETAILS\n")

            for room_number, details in rooms.items():

                file.write(f"\nRoom Number: {room_number}\n")
                file.write(f"Room Type: {details['type']}\n")
                file.write(f"Price: ${details['price']}\n")
                file.write(f"Features: {details['features']}\n")
                file.write(f"Status: {details['status']}\n")

            file.write("\nCURRENT ROOM ALLOCATIONS\n")

            for room_number, details in allocations.items():

                file.write(f"\nRoom Number: {room_number}\n")
                file.write(f"Customer: {details['customer']}\n")
                file.write(f"Nights: {details['nights']}\n")
                file.write(f"Date: {details['date']}\n")

        print("Room allocation data saved successfully!")
        print("File Name:", filename)

    except Exception as e:
        print("Error saving file:", e)


# MENU OPTION 8 - DISPLAY ROOM ALLOCATION FROM FILE

def display_from_file():
    print("\n--- DISPLAY DATA FROM FILE ---")

    try:
        if not os.path.exists(filename):
            print("File does not exist. Please save data first.")
            return

        with open(filename, "r") as file:
            content = file.read()

        if content:
            print("\n--- SAVED HOTEL DATA ---")
            print(content)

        else:
            print("The file is empty.")

    except Exception as e:
        print("Error reading file:", e)


# MENU OPTION 9 - BACKUP AND CLEAR FILE

def backup_file():
    print("\n--- BACKUP AND CLEAR FILE ---")

    try:
        if not os.path.exists(filename):
            print("File does not exist!")
            return

        with open(filename, "r") as file:
            content = file.read()

        if not content:
            print("File is empty. Nothing to backup.")
            return

        date_time = datetime.now().strftime("%Y%m%d_%H%M%S")

        backup_filename = (
            f"LHMS_850011021_Backup_{date_time}.txt"
        )

        with open(backup_filename, "a") as backup:
            backup.write(content)
            backup.write("\n")

        with open(filename, "w") as file:
            file.write("")

        print("Backup created successfully!")
        print("Backup File:", backup_filename)
        print("Original file content cleared successfully!")

    except Exception as e:
        print("Error during backup:", e)


# MAIN APPLICATION MENU

while True:

    print("\n========================================")
    print("       LANGHAM HOTEL MANAGEMENT SYSTEM")
    print("========================================")

    print("1. Add Room")
    print("2. Delete Room")
    print("3. Display Room Details")
    print("4. Allocate Room")
    print("5. Display Room Allocation Details")
    print("6. Billing and De-Allocation")
    print("7. Save Room Allocation to File")
    print("8. Display Room Allocation from File")
    print("9. Backup and Clear Room Allocation File")
    print("0. Exit Application")

    print("========================================")

    try:
        choice = int(input("Enter Your Choice (0-9): "))

        if choice == 1:
            add_room()

        elif choice == 2:
            delete_room()

        elif choice == 3:
            display_rooms()

        elif choice == 4:
            allocate_room()

        elif choice == 5:
            display_allocations()

        elif choice == 6:
            billing()

        elif choice == 7:
            save_to_file()

        elif choice == 8:
            display_from_file()

        elif choice == 9:
            backup_file()

        elif choice == 0:
            print("\nThank you for using Langham Hotel Management System!")
            print("Application closed.")
            break

        else:
            print("Invalid choice! Please enter a number between 0 and 9.")

    except ValueError:
        print("Invalid input! Please enter a valid number.")

    except Exception as e:
        print("An unexpected error occurred:", e)
```
