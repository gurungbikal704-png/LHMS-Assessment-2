```python
# LANGHAM HOTEL MANAGEMENT SYSTEM
# TASK 1 AND TASK 2
# Error Detection and Exception Handling

import os
import math
from datetime import datetime

# Initialising dictionaries
rooms = {}
allocations = {}

# File name
filename = "LHMS_850011021.txt"


# INPUT VALIDATION FUNCTIONS

def get_number(prompt):
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            print("Error: Please enter a valid whole number.")
        except EOFError:
            print("Error: No input received. Please try again.")


def get_price(prompt):
    while True:
        try:
            price = float(input(prompt))

            if not math.isfinite(price) or price <= 0:
                print("Price must be a positive finite number.")
                continue

            return price

        except ValueError:
            print("Error: Enter a valid price.")
        except OverflowError:
            print("Error: Price is too large.")
        except EOFError:
            print("Error: No input received.")


# MENU OPTION 1 - ADD ROOM

def add_room():
    print("\n--- ADD ROOM ---")

    room_number = input("Enter Room Number: ")

    if room_number in rooms:
        print("Room already exists!")
        return

    room_type = input("Enter Room Type: ")
    price = get_price("Enter Price Per Night: ")
    features = input("Enter Room Features: ")

    rooms[room_number] = {
        "type": room_type,
        "price": price,
        "features": features,
        "status": "Available"
    }

    print("Room added successfully!")


# MENU OPTION 2 - DELETE ROOM

def delete_room():
    print("\n--- DELETE ROOM ---")

    room_number = input("Enter Room Number to Delete: ")

    if room_number not in rooms:
        print("Room not found!")
        return

    if rooms[room_number]["status"] == "Allocated":
        print("Cannot delete an allocated room.")
        return

    del rooms[room_number]

    print("Room deleted successfully!")


# MENU OPTION 3 - DISPLAY ROOM DETAILS

def display_rooms():
    print("\n--- ROOM DETAILS ---")

    if not rooms:
        print("No rooms available.")
        return

    for room_number, details in rooms.items():

        print("\nRoom Number:", room_number)
        print("Room Type:", details["type"])
        print("Price Per Night: $", details["price"])
        print("Features:", details["features"])
        print("Status:", details["status"])


# MENU OPTION 4 - ALLOCATE ROOM

def allocate_room():
    print("\n--- ALLOCATE ROOM ---")

    room_number = input("Enter Room Number: ")

    if room_number not in rooms:
        print("Room does not exist!")
        return

    if rooms[room_number]["status"] == "Allocated":
        print("Room is already allocated!")
        return

    customer_name = input("Enter Customer Name: ")

    nights = get_number("Enter Number of Nights: ")

    if nights <= 0:
        print("Number of nights must be greater than zero.")
        return

    allocations[room_number] = {
        "customer": customer_name,
        "nights": nights,
        "date": datetime.now().strftime("%Y-%m-%d")
    }

    rooms[room_number]["status"] = "Allocated"

    print("Room allocated successfully!")


# MENU OPTION 5 - DISPLAY ALLOCATION DETAILS

def display_allocations():
    print("\n--- ROOM ALLOCATION DETAILS ---")

    if not allocations:
        print("No rooms are currently allocated.")
        return

    for room_number, details in allocations.items():

        print("\nRoom Number:", room_number)
        print("Customer Name:", details["customer"])
        print("Number of Nights:", details["nights"])
        print("Allocation Date:", details["date"])
        print("Room Status:", rooms[room_number]["status"])


# MENU OPTION 6 - BILLING AND DE-ALLOCATION

def billing():
    print("\n--- BILLING AND DE-ALLOCATION ---")

    room_number = input("Enter Room Number: ")

    if room_number not in allocations:
        print("Room is not currently allocated!")
        return

    customer = allocations[room_number]["customer"]
    nights = allocations[room_number]["nights"]
    price = rooms[room_number]["price"]

    total_bill = nights * price

    print("\n--- CUSTOMER BILL ---")
    print("Customer Name:", customer)
    print("Room Number:", room_number)
    print("Number of Nights:", nights)
    print("Price Per Night: $", price)
    print("Total Bill: $", total_bill)

    print("\nPayment completed.")

    del allocations[room_number]

    rooms[room_number]["status"] = "Available"

    print("Room de-allocated successfully!")
    print("Room is now available.")


# MENU OPTION 7 - SAVE DATA TO FILE

def save_to_file():
    print("\n--- SAVE ROOM ALLOCATION ---")

    try:
        with open(filename, "w") as file:

            file.write("LANGHAM HOTEL MANAGEMENT SYSTEM\n")
            file.write("=" * 40 + "\n")

            file.write("\nROOM DETAILS\n")

            for room_number, details in rooms.items():

                file.write(f"\nRoom Number: {room_number}\n")
                file.write(f"Room Type: {details['type']}\n")
                file.write(f"Price: ${details['price']}\n")
                file.write(f"Features: {details['features']}\n")
                file.write(f"Status: {details['status']}\n")

            file.write("\nCURRENT ALLOCATIONS\n")

            for room_number, details in allocations.items():

                file.write(f"\nRoom Number: {room_number}\n")
                file.write(f"Customer: {details['customer']}\n")
                file.write(f"Nights: {details['nights']}\n")
                file.write(f"Date: {details['date']}\n")

        print("Data saved successfully!")

    except OSError as e:
        print("File saving error:", e)


# MENU OPTION 8 - DISPLAY DATA FROM FILE

def display_from_file():
    print("\n--- DISPLAY DATA FROM FILE ---")

    try:
        with open(filename, "r") as file:
            content = file.read()

        if content:
            print(content)
        else:
            print("The file is empty.")

    except FileNotFoundError:
        print("Error: File does not exist.")

    except OSError as e:
        print("File reading error:", e)


# MENU OPTION 9 - BACKUP AND CLEAR FILE

def backup_file():
    print("\n--- BACKUP AND CLEAR FILE ---")

    try:
        with open(filename, "r") as file:
            content = file.read()

        if not content:
            print("File is empty. Nothing to backup.")
            return

        date_time = datetime.now().strftime("%Y%m%d_%H%M%S")

        backup_filename = (
            f"LHMS_850011021_Backup_{date_time}.txt"
        )

        with open(backup_filename, "a") as backup:
            backup.write(content)
            backup.write("\n")

        with open(filename, "w") as file:
            file.write("")

        print("Backup created successfully!")
        print("Backup File:", backup_filename)
        print("Original file cleared successfully!")

    except FileNotFoundError:
        print("Error: Original file does not exist.")

    except OSError as e:
        print("Backup error:", e)


# TASK 2 - EXCEPTION HANDLING DEMONSTRATION

def demonstrate_exceptions():

    print("\n--- EXCEPTION HANDLING DEMONSTRATION ---")

    # 1. SyntaxError
    try:
        compile("if True print('Hello')", "<test>", "exec")
    except SyntaxError:
        print("1. SyntaxError handled successfully.")

    # 2. ValueError
    try:
        int("Hello")
    except ValueError:
        print("2. ValueError handled successfully.")

    # 3. ZeroDivisionError
    try:
        result = 10 / 0
    except ZeroDivisionError:
        print("3. ZeroDivisionError handled successfully.")

    # 4. IndexError
    try:
        sample_list = [10, 20, 30]
        print(sample_list[5])
    except IndexError:
        print("4. IndexError handled successfully.")

    # 5. NameError
    try:
        print(undefined_variable)
    except NameError:
        print("5. NameError handled successfully.")

    # 6. TypeError
    try:
        result = "Hotel" + 100
    except TypeError:
        print("6. TypeError handled successfully.")

    # 7. OverflowError
    try:
        result = math.exp(1000)
    except OverflowError:
        print("7. OverflowError handled successfully.")

    # 8. IOError
    try:
        raise IOError("Simulated file input/output error")
    except IOError:
        print("8. IOError handled successfully.")

    # 9. ImportError
    try:
        import nonexistent_hotel_module
    except ImportError:
        print("9. ImportError handled successfully.")

    # 10. EOFError
    try:
        raise EOFError("Simulated end of input")
    except EOFError:
        print("10. EOFError handled successfully.")

    # 11. FileNotFoundError
    try:
        with open("missing_hotel_file.txt", "r") as file:
            content = file.read()
    except FileNotFoundError:
        print("11. FileNotFoundError handled successfully.")

    print("\nAll 11 exception demonstrations completed!")


# MAIN APPLICATION MENU

while True:

    print("\n========================================")
    print("    LANGHAM HOTEL MANAGEMENT SYSTEM")
    print("========================================")

    print("1. Add Room")
    print("2. Delete Room")
    print("3. Display Room Details")
    print("4. Allocate Room")
    print("5. Display Room Allocation Details")
    print("6. Billing and De-Allocation")
    print("7. Save Room Allocation to File")
    print("8. Display Room Allocation from File")
    print("9. Backup and Clear Room Allocation File")
    print("10. Demonstrate Exception Handling")
    print("0. Exit Application")

    print("========================================")

    try:

        choice = get_number("Enter Your Choice: ")

        if choice == 1:
            add_room()

        elif choice == 2:
            delete_room()

        elif choice == 3:
            display_rooms()

        elif choice == 4:
            allocate_room()

        elif choice == 5:
            display_allocations()

        elif choice == 6:
            billing()

        elif choice == 7:
            save_to_file()

        elif choice == 8:
            display_from_file()

        elif choice == 9:
            backup_file()

        elif choice == 10:
            demonstrate_exceptions()

        elif choice == 0:
            print("\nThank you for using Langham Hotel!")
            print("Application closed.")
            break

        else:
            print("Invalid choice! Enter a number from 0 to 10.")

    except KeyboardInterrupt:
        print("\nApplication interrupted by user.")
        break

    except Exception as e:
        print("Unexpected error:", e)
```
